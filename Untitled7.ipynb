```python
import pandas as pd
import numpy as np
import random
from datetime import datetime, timedelta

# Data size - Pehle 10,000 records generate kar rahe hain
num_records = 10000

# Constants for simulation
platforms = ['Facebook', 'Instagram', 'Google', 'TikTok', 'YouTube']
campaigns = ['Summer_Sale', 'Winter_Promo', 'Flash_Sale', 'Brand_Awareness']
genders = ['Male', 'Female', 'Other']
cities = ['Karachi', 'Lahore', 'Islamabad', 'Peshawar', 'Quetta']
devices = ['Mobile', 'Desktop', 'Tablet']

# Dates generate karna (pichle 30 din ka data)
start_date = datetime.today() - timedelta(days=30)
dates = [start_date + timedelta(days=random.randint(0, 30)) for _ in range(num_records)]

# Random data generate karna
data = {
    'date': dates,
    'campaign': np.random.choice(campaigns, num_records),
    'platform': np.random.choice(platforms, num_records),
    'age': np.random.randint(18, 65, num_records),
    'gender': np.random.choice(genders, num_records),
    'city': np.random.choice(cities, num_records),
    'device': np.random.choice(devices, num_records),
}

df = pd.DataFrame(data)

# Metrics simulate karna (logical limits ke sath)
df['impressions'] = np.random.randint(1000, 50000, num_records)
# Clicks humesha impressions se kam honge
df['clicks'] = df['impressions'] * np.random.uniform(0.01, 0.10, num_records)
df['clicks'] = df['clicks'].astype(int)
# Spend
df['spend'] = df['clicks'] * np.random.uniform(0.5, 5.0, num_records)
df['spend'] = df['spend'].round(2)
# Conversions clicks ka kuch percentage honge
df['conversions'] = df['clicks'] * np.random.uniform(0.01, 0.20, num_records)
df['conversions'] = df['conversions'].astype(int)
# Revenue conversions par depend karega
df['revenue'] = df['conversions'] * np.random.uniform(20.0, 150.0, num_records)
df['revenue'] = df['revenue'].round(2)

# File ko CSV mein save karna taake aage use kar sakein
file_name = 'digital_marketing_data_10k.csv'
df.to_csv(file_name, index=False)

print(f"Data generation complete! File saved as {file_name}")
print("\nDataset ki pehli 5 lines:\n")
print(df.head())
```

```python
import pandas as pd
import numpy as np

# Colab mein jo file pehle bani thi usay load karna
df = pd.read_csv('digital_marketing_data_10k.csv')

# 1. CTR (Click-Through Rate) = (Clicks / Impressions) * 100
df['CTR_%'] = (df['clicks'] / df['impressions']) * 100

# 2. CPC (Cost Per Click) = Spend / Clicks
df['CPC'] = np.where(df['clicks'] > 0, df['spend'] / df['clicks'], 0)

# 3. Conversion Rate = (Conversions / Clicks) * 100
df['Conversion_Rate_%'] = np.where(df['clicks'] > 0, (df['conversions'] / df['clicks']) * 100, 0)

# 4. CPA (Cost Per Acquisition) = Spend / Conversions
df['CPA'] = np.where(df['conversions'] > 0, df['spend'] / df['conversions'], 0)

# 5. ROAS (Return On Ad Spend) = Revenue / Spend
df['ROAS'] = np.where(df['spend'] > 0, df['revenue'] / df['spend'], 0)

# Decimals ko 2 tak mehdood karna taake data saaf dikhe
df = df.round({'CTR_%': 2, 'CPC': 2, 'Conversion_Rate_%': 2, 'CPA': 2, 'ROAS': 2})

# Metrics wali nayi file save karna
file_name = 'digital_marketing_metrics_10k.csv'
df.to_csv(file_name, index=False)

print(f"Metrics successfully calculate ho gaye hain! Nayi file '{file_name}' ban gayi hai.")
print("\nData ki pehli 5 lines nayi calculations ke sath:\n")
# Sirf zaroori columns dikha rahe hain
print(df[['platform', 'campaign', 'CTR_%', 'CPC', 'Conversion_Rate_%', 'CPA', 'ROAS']].head())
```

```python
import pandas as pd
import numpy as np
import random
from datetime import datetime, timedelta

# Data generate karne ka function
def generate_marketing_data(num_records):
    print(f"{num_records} records generate ho rahe hain. Kripya thoda wait karein...")

    platforms = ['Facebook', 'Instagram', 'Google', 'TikTok', 'YouTube']
    campaigns = ['Summer_Sale', 'Winter_Promo', 'Flash_Sale', 'Brand_Awareness']
    genders = ['Male', 'Female', 'Other']
    cities = ['Karachi', 'Lahore', 'Islamabad', 'Peshawar', 'Quetta']
    devices = ['Mobile', 'Desktop', 'Tablet']

    start_date = datetime.today() - timedelta(days=30)

    df = pd.DataFrame({
        'date': [start_date + timedelta(days=random.randint(0, 30)) for _ in range(num_records)],
        'campaign': np.random.choice(campaigns, num_records),
        'platform': np.random.choice(platforms, num_records),
        'age': np.random.randint(18, 65, num_records),
        'gender': np.random.choice(genders, num_records),
        'city': np.random.choice(cities, num_records),
        'device': np.random.choice(devices, num_records),
    })

    df['impressions'] = np.random.randint(1000, 50000, num_records)
    df['clicks'] = (df['impressions'] * np.random.uniform(0.01, 0.10, num_records)).astype(int)
    df['spend'] = (df['clicks'] * np.random.uniform(0.5, 5.0, num_records)).round(2)
    df['conversions'] = (df['clicks'] * np.random.uniform(0.01, 0.20, num_records)).astype(int)
    df['revenue'] = (df['conversions'] * np.random.uniform(20.0, 150.0, num_records)).round(2)

    file_name = f'digital_marketing_data_{num_records}.csv'
    df.to_csv(file_name, index=False)
    print(f"Data ready! File saved as {file_name}")
    return df

# 1. 100,000 records generate karein
df_100k = generate_marketing_data(100000)

# 2. 1,000,000 (1 Million) records generate karein
# (1 million mein thoda zyada time lag sakta hai)
df_1m = generate_marketing_data(1000000)
```

```python
import pandas as pd

print("1 Million records ka data load ho raha hai...")
# 1 Million wala data load karein
df = pd.read_csv('digital_marketing_data_1000000.csv')

# Har platform ka total spend aur total revenue calculate karein
platform_stats = df.groupby('platform')[['spend', 'revenue', 'conversions']].sum().reset_index()

# Platform ka overall ROAS (Return On Ad Spend) nikalein
platform_stats['ROAS'] = platform_stats['revenue'] / platform_stats['spend']

# ROAS ke hisab se budget ka percentage nikalein
# Jiski performance (ROAS) sabse achi, usko sabse zyada percentage milega
total_roas = platform_stats['ROAS'].sum()
platform_stats['budget_percentage'] = (platform_stats['ROAS'] / total_roas) * 100

# Rs. 1 Million (10,000,000) ka naya budget allocate karein
total_new_budget = 1000000
platform_stats['new_budget_allocation'] = (platform_stats['budget_percentage'] / 100) * total_new_budget

# Decimals ko round karein taake result saaf dikhe
platform_stats['budget_percentage'] = platform_stats['budget_percentage'].round(2)
platform_stats['new_budget_allocation'] = platform_stats['new_budget_allocation'].round(2)

# Results ko ROAS ke hisab se highest to lowest sort karein
platform_stats = platform_stats.sort_values(by='ROAS', ascending=False).reset_index(drop=True)

print("\n=== MARKETING BUDGET RECOMMENDATION ENGINE ===")
print("Company ko agla Rs. 1 Million is tarah spend karna chahiye:\n")

for index, row in platform_stats.iterrows():
    print(f"{row['platform']}: {row['budget_percentage']}% (Rs. {row['new_budget_allocation']:,.2f})")

print("\nDetail Data:")
print(platform_stats[['platform', 'ROAS', 'budget_percentage', 'new_budget_allocation']])
```

```python
# Sab se pehle website banane wali library install karein
!pip install gradio

import gradio as gr
import pandas as pd
import matplotlib.pyplot as plt

# Yeh function pichli performance (ROAS) ke hisab se naya budget calculate karega
def calculate_budget(new_budget):
    # Platforms aur unki pichli performance (ROAS) ka data
    data = {
        'Platform': ['TikTok', 'Instagram', 'Facebook', 'Google', 'YouTube'],
        'ROAS': [6.0, 5.0, 4.0, 3.0, 2.5]
    }
    df = pd.DataFrame(data)

    # Percentage aur naya budget calculate karna
    total_roas = df['ROAS'].sum()
    df['Percentage (%)'] = (df['ROAS'] / total_roas) * 100
    df['Allocated Budget (Rs)'] = (df['Percentage (%)'] / 100) * new_budget

    # Decimals ko set karna aur sort karna
    df['Percentage (%)'] = df['Percentage (%)'].round(2)
    df['Allocated Budget (Rs)'] = df['Allocated Budget (Rs)'].round(2)
    df = df.sort_values(by='Percentage (%)', ascending=False)

    # Khoobsurat Chart banana
    fig, ax = plt.subplots(figsize=(8, 5))
    colors = ['#000000', '#E1306C', '#3b5998', '#DB4437', '#FF0000']
    ax.bar(df['Platform'], df['Allocated Budget (Rs)'], color=colors)
    ax.set_title(f'Rs. {new_budget:,.0f} Budget Allocation', fontsize=14, fontweight='bold')
    ax.set_ylabel('Amount (Rs)')
    plt.tight_layout()

    return df, fig

# Website ka user interface (UI) design karna
app = gr.Interface(
    fn=calculate_budget,
    inputs=gr.Number(value=1000000, label="Company ka Naya Budget Enter Karein (Jaise: 1000000)"),
    outputs=[
        gr.Dataframe(label="Recommendation Table"),
        gr.Plot(label="Budget Allocation Chart")
    ],
    title="📊 Digital Advertising Budget Optimizer",
    description="Apna budget yahan enter karein. Yeh AI Recommendation Engine pichli performance ke hisab se automatically batayega ke kis platform par kitne paise lagane chahiye."
)

# Website ko internet par live karna
app.launch(share=True)
```

```python
# Zaroori libraries install karein
!pip install gradio pandas matplotlib

import gradio as gr
import pandas as pd
import matplotlib.pyplot as plt

# Budget aur data calculate karne ka function
def calculate_budget(new_budget):
    if new_budget <= 0:
        return pd.DataFrame(), plt.figure()

    # Pichli performance ka advanced data (ROAS aur Conversion Rate)
    data = {
        'Platform': ['TikTok', 'Instagram', 'Facebook', 'Google', 'YouTube'],
        'Historical ROAS': [6.5, 5.2, 4.0, 3.1, 2.5],
        'Conversion Rate (%)': [4.5, 3.8, 3.0, 2.5, 1.8]
    }
    df = pd.DataFrame(data)

    # Naya budget aur percentage nikalna
    total_roas = df['Historical ROAS'].sum()
    df['Allocation (%)'] = (df['Historical ROAS'] / total_roas) * 100
    df['Suggested Budget (Rs)'] = (df['Allocation (%)'] / 100) * new_budget

    # Numbers ko round off karna
    df['Allocation (%)'] = df['Allocation (%)'].round(2)
    df['Suggested Budget (Rs)'] = df['Suggested Budget (Rs)'].round(2)
    df = df.sort_values(by='Allocation (%)', ascending=False)

    # Professional aur modern chart design karna
    fig, ax = plt.subplots(figsize=(10, 5))
    colors = ['#000000', '#E1306C', '#1877F2', '#DB4437', '#FF0000']

    bars = ax.bar(df['Platform'], df['Suggested Budget (Rs)'], color=colors, edgecolor='black', alpha=0.85)

    # Bars ke upar exact amount likhna
    for bar in bars:
        yval = bar.get_height()
        ax.text(bar.get_x() + bar.get_width()/2, yval + (new_budget*0.02), f'Rs {yval:,.0f}',
                ha='center', va='bottom', fontsize=10, fontweight='bold')

    ax.set_title('📊 Optimal Budget Allocation Strategy', fontsize=16, pad=15, fontweight='bold', color='#333')
    ax.set_ylabel('Allocated Amount (PKR)', fontsize=12, fontweight='bold')
    ax.grid(axis='y', linestyle='--', alpha=0.5)
    ax.spines['top'].set_visible(False)
    ax.spines['right'].set_visible(False)
    plt.tight_layout()

    return df, fig

# Professional Dashboard Interface ban banana
with gr.Blocks(theme=gr.themes.Soft(primary_hue="indigo", secondary_hue="blue")) as app:

    # Website Header
    gr.Markdown(
        """
        # 🚀 Digital Advertising Budget Optimizer
        **Executive Dashboard:** Enter your total marketing budget below. Our algorithm analyzes simulated historical campaign performance (CTR, CPC, CPA, ROAS) to suggest the most profitable budget allocation across major digital platforms.
        """
    )

    # Columns aur Rows ka Layout
    with gr.Row():
        # Left Column: Input ke liye
        with gr.Column(scale=1):
            gr.Markdown("### 💰 Budget Configuration")
            budget_input = gr.Number(
                value=1000000,
                label="Total Marketing Budget (Rs)",
                info="Example: 1000000 (1 Million)",
                interactive=True
            )
            calculate_btn = gr.Button("🧠 Generate Strategy", variant="primary")

        # Right Column: Data Table ke liye
        with gr.Column(scale=2):
            gr.Markdown("### 📈 Recommendation Engine Output")
            output_table = gr.Dataframe(
                label="Detailed Budget Allocation",
                headers=["Platform", "Historical ROAS", "Conversion Rate (%)", "Allocation (%)", "Suggested Budget (Rs)"]
            )

    # Bottom Row: Chart ke liye
    with gr.Row():
        output_plot = gr.Plot(label="Visual Budget Distribution")

    # Button click hone par function run karna
    calculate_btn.click(
        fn=calculate_budget,
        inputs=budget_input,
        outputs=[output_table, output_plot]
    )

# Aakhir mein link generate karne ka code
print("Website ban rahi hai. Niche diye gaye '.gradio.live' wale link par click karein:")
app.launch(share=True)
```
